```python
import math

def minimax(depth, node_index, is_maximizing_player, values, max_depth):
    
    if depth == max_depth:
        return values[node_index]
    
    
    if is_maximizing_player:
        best_value = max(
            minimax(depth + 1, node_index * 2,     False, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, False, values, max_depth)
        )
        return best_value
    
    
    else:
        best_value = min(
            minimax(depth + 1, node_index * 2,     True, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, True, values, max_depth)
        )
        return best_value



scores = [3, 5, 2, 9, 12, 5, 23, 15]  


tree_depth = int(math.log2(len(scores)))  

print("Leaf nodes ke values:", scores)
print("Tree ki depth:", tree_depth)
print("Optimal value (jo player choose karega):", 
      minimax(0, 0, True, scores, tree_depth))
```
