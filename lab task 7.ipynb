```python
class RouteFinder:
    def __init__(self, graph_data):
        self.graph_data = graph_data

    def neighbors_of(self, node):
        return self.graph_data.get(node, [])

    def heuristic(self, node):
        return 1  

    def build_path(self, came_from, start, goal):
        route = []
        current = goal
        while current != start:
            route.append(current)
            current = came_from[current]
        route.append(start)
        route.reverse()
        return route

    def a_star_search(self, start, goal):
        open_set = {start}
        closed_set = set()
        cost = {start: 0}
        came_from = {start: None}

        while open_set:
            current = min(open_set, key=lambda x: cost[x] + self.heuristic(x))

            if current == goal:
                path = self.build_path(came_from, start, goal)
                print("Path found:", " → ".join(path))
                print("Steps taken:", len(path) - 1)
                print("Total path cost:", cost[goal])
                return path

            for neighbor, distance in self.neighbors_of(current):
                new_cost = cost[current] + distance
                if neighbor not in cost or new_cost < cost[neighbor]:
                    cost[neighbor] = new_cost
                    came_from[neighbor] = current
                    if neighbor not in closed_set:
                        open_set.add(neighbor)

            open_set.remove(current)
            closed_set.add(current)

        print("No route found between", start, "and", goal)
        return None


routes = {
    'P': [('Q', 1), ('R', 3), ('S', 7)],
    'Q': [('S', 5)],
    'R': [('S', 12)],
    'S': []
}

finder = RouteFinder(routes)
finder.a_star_search('P', 'S')
```
