```python
 
print(" SPLITTING INTO X AND Y:")
target_column = df.columns[-1]
print(f"Target column: '{target_column}'")
x = df.iloc[:, 0:-1] 
y = df.iloc[:, -1]    

print(f"X shape: {x.shape}")
print(f"y shape: {y.shape}")
print(f"X columns: {list(x.columns)}")
print(f"y column: {target_column}")
```

```python

print(" CONVERTING CATEGORICAL COLUMNS TO INTEGERS:")
cat_columns = x.select_dtypes(['object']).columns
print(f"Categorical columns found: {list(cat_columns)}")

if len(cat_columns) > 0:
    x[cat_columns] = x[cat_columns].apply(lambda col: pd.factorize(col)[0])
    print(" Categorical columns converted to integers using factorize")
    print("\nCategorical Column Mappings:")
    for col in cat_columns:
        unique_vals = df[col].unique()
        mapping = {i: val for i, val in enumerate(unique_vals)}
        print(f"{col}: {mapping}")
else:
    print(" No categorical columns found in ")
```
